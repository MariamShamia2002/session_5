```python
def classify_game(turns):
    if turns < 20:
        return 'Short'
    elif turns < 50:
        return 'Medium'
    else:
        return 'Long'
df['game_length'] = df['turns'].apply(classify_game)
df['game_length'].value_counts(normalize=True) * 100
```
